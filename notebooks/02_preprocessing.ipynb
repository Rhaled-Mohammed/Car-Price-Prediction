```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("../data/car_price_prediction.csv")
```

```python
#Duplicate rows check

#number of duplicates is 313 wich is not a lot but still considerable, so we will drop them to avoid any bias in the model.

data = data.drop_duplicates()

```

```python
#Data conversion
data["Levy"] = data["Levy"].replace("-",0)
data["Levy"] = pd.to_numeric(data["Levy"])


data["Engine volume"] = data["Engine volume"].str.replace("Turbo","")
data["Engine volume"] = pd.to_numeric(data["Engine volume"])


data["Mileage"] = data["Mileage"].str.replace("km",'')
data["Mileage"]= pd.to_numeric(data["Mileage"])

#Converted Levy, Engine volume and Mileage columns to numeric by removing str parts.
```

```python
#Outliers
def remove_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 3 * IQR
    upper_bound = Q3 + 3 * IQR
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

data = remove_outliers(data, "Levy")
data = remove_outliers(data, "Mileage")
data = remove_outliers(data, "Price")

#We removed outliers from Levy, Mileage and Price columns using the extreme IQR method wich contains the same outliers as normal method. This will help to improve the model's performance by reducing the influence of extreme values.
```

```python


data.to_csv("../data/car_price_prediction_cleaned.csv", index=False)
```
